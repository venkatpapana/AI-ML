### Dot Product Example (Monthly Prints)

```python
monthly_prints = pd.read_csv("data/monthly_prints.csv", index_col=None)
monthly_prints
```

```python
monthly_prints.set_index("Month", inplace=True)
monthly_prints
```

```python
num_plies = pd.DataFrame([[1, 2, 3, 4]], columns=["F&I", "Parts", "Service", "Accounting"], index=["Num_Plies"])
num_plies
```

```python
monthly_prints.shape, num_plies.shape
```

```python
 num_plies.T.shape, monthly_prints.shape
```

```python
Total_Plies_Printed_Per_Dept = monthly_prints.dot(num_plies.T)
Total_Plies_Printed_Per_Dept
```

```python
monthly_prints_result = monthly_prints
monthly_prints_result["Total Plies"] = Total_Plies_Printed_Per_Dept
monthly_prints_result
```

```python
[colname for colname in monthly_prints_result.columns]
```

```python
monthly_prints_result.loc["Total"] = [monthly_prints_result[colname].sum() for colname in monthly_prints_result.columns]
```

```python
monthly_prints_result
```

```python
monthly_prints_result.to_csv("data/monthly_prints_exported.csv")
```
